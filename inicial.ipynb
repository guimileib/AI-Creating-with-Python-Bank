### Case: Score de Crédito dos Clientes

Você foi contratado por um banco para conseguir definir o score de crédito dos clientes. Você precisa analisar todos os clientes do banco e, com base nessa análise, criar um modelo que consiga ler as informações do cliente e dizer automaticamente o score de crédito dele: Ruim, Ok, Bom

Arquivos da aula: https://drive.google.com/drive/folders/1FbDqVq4XLvU85VBlVIMJ73p9oOu6u2-J?usp=drive_link

```python
'''
Passo 0: Entender o desafio e a empresa
Tenho que avaliar o Score de crédito do cliente e dizer se:
    - É bom - good 
    - É ruim - poor
    - É ok - standard
'''

'''
    Passo a passo:
    - Passo 0: Entender o desafio e a empresa
    - Passo 1: Importar a base de dados
    - Passo 2: Preparar a base de dados pra IA - Toda IA é preciso preparar os dados para a IA
    - Passo 3: Cria modelo de IA => prever a nota de crédito
    - Passo 4: Escolher o melhor modelo
    - Passo 5: Fazer novas previsões
'''
```

```python
# Passo 1: Importar a base de dados
import pandas as pd

tabela = pd.read_csv('clientes.csv') 

display(tabela)
```

```python
# Passo 2: Preparar a base de dados pra IA - Toda IA é preciso preparar os dados para a IA
'''
    Olhar a base de dados toda e ela vai aprender o que é um cliente e o que eu quero prever do cliente
    que nesse caso é prever a nota de credito do cliente
'''

'''
Anotando as colunas que estão em text(object) - Esse processo se chama LabelEncoder
    - 
    - mix_credito   
    - comportamento_pagamento             
'''
# Label Encoder - explicação está no Notion
from sklearn.preprocessing import LabelEncoder

# profissao
codificador1 = LabelEncoder()
tabela['profissao'] = codificador1.fit_transform(tabela['profissao']) 

'''
o novo tabela da minha profissao (tabela['profissao'])  
vai ser o antigo valor da tabela profissão aplicando o codificador nele: 
codificador1.fit_transform(tabela['profissao']) 
'''

# mix_credito
codificador2 = LabelEncoder()
tabela['mix_credito'] = codificador2.fit_transform(tabela['mix_credito'])

#comportamento_pagamento
codificador3 = LabelEncoder()
tabela['comportamento_pagamento'] = codificador3.fit_transform(tabela['comportamento_pagamento'])

display(tabela.info()) # info das colunas


```

```python
# Separar as informações da base de dados para a IA

# Separar a base de dados em X e Y
# Y -> quem eu quero prever
# X -> as outras colunas (todas as colunas que eu vou usar pra fazer a previsão)
y = tabela['score_credito']
x = tabela.drop(columns=['score_credito', 'id_cliente']) 

# Separar em dados de treino e de teste
from sklearn.model_selection import train_test_split

x_training, x_test, y_training, y_test = train_test_split(x, y, test_size=0.3) # A ordem tem que ser essa

```

```python
# Passo 3: Criar modelo de IA => prever a nota de crédito
'''
    1. importar o modelo
    2. criar modelo
    3. treinar o modelo
'''
# importar o modelo
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# criar o modelo
modelo_arvoredecisao = RandomForestClassifier()
modelo_knn = KNeighborsClassifier()

# treinar o modelo
modelo_arvoredecisao.fit(x_training, y_training)
modelo_knn.fit(x_training, y_training)
```

```python
# Passo 4: Escolher o melhor modelo

# previsões 
previsao_arvoredecisao = modelo_arvoredecisao.predict(x_test)
previsao_knn = modelo_knn.predict(x_test)

# acurácia- exatidão ou proximidade de um resultado com um valor de referência
from sklearn.metrics import accuracy_score

display(accuracy_score(y_test, previsao_arvoredecisao)) # foi o melhor modelo, conforme a precisao
display(accuracy_score(y_test, previsao_knn))
```

```python
# Passo 5: Fazer novas previsões
# o melhor método modelo é o árvore de decisão

# 
tabela_nova = pd.read_csv('novos_clientes.csv')

# profissao
codificador1 = LabelEncoder()
tabela_nova['profissao'] = codificador1.fit_transform(tabela_nova['profissao']) 

# mix_credito
codificador2 = LabelEncoder()
tabela_nova['mix_credito'] = codificador2.fit_transform(tabela_nova['mix_credito'])

#comportamento_pagamento
codificador3 = LabelEncoder()
tabela_nova['comportamento_pagamento'] = codificador3.fit_transform(tabela_nova['comportamento_pagamento'])

display(tabela_nova)

previsao = modelo_arvoredecisao.predict(tabela_nova)
display(previsao)
```
